## Sorting values

```python
#sorting by ascending
df_exams.count_values(['math score', 'reading score'])
#sorting my descending
df_exams.count_values(['math score', 'reading score'], ascending=False, inplace=True)
# inplace=True to update data frame, as default =False (doesn't change)
# you can also overwrite the argument rather than using the inplace option
```

```python
# sort descending with a key function
df_exams.count_values('race/ ethnicity', ascending=True)
```

## Index

```python
import numpy as np
import random
```

```python
# creating non repetitive values for the index
new_index = np.arrange(0, 1000)
```

```python
# shuffling the indexes
random.shuffle(new_index)
```

```python
# creating a new column with new index
df_exams['new_index'] = new_index
df_exams
```

```python
# set new_index column as index
df_exams.set_index['new_index', inplace=True]
```

```python
# sort dataframe by new index
df_exams.sort_index()
#sort descending
df_exams.sort_index(ascending=False, inplace=True)
# inplace makes changes permanent
```

## Rename Column

```python
import pandas as pd
```

```python
df_exams = pd.read_csv(' ')
df_exams.head()
```

```python
#rename columns
df_exams = df_exams.rename(columns={'gender':'Gender'})
```
