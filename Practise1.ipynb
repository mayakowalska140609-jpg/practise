## If statement

## For loop

```python
for <variable> in <list> :
    <code>
```

```python
names = ["Jeremy", "Steward", "Bob", "Pocahontas"]
```

```python
names
```

```python
for i, name in enumerate (names) :
    print (i)
    print (name)
```

```python
my_data = {'name':'Maya', 'age':'17'}
```

```python
for key, value in my_data.items() :
    print (key)
    print (value)
```
